```python
training_data = [[1, 2], [2, 3], [3, 4], [6, 7], [7, 8],[8,9]]
training_labels = ['A', 'A','c','b','c','b']
test_point = [4, 5]
k = 3
```

```python
import numpy as np
from collections import Counter
```

```python
def euclidean_distance(p1,p2):
    return np.sqrt(np.sum((np.array(p1)-np.array(p2))**2))

def knn_predict(training_Data, training_labels, test_point, k):
    ds=[]
    for i in range(len(training_Data)):
        dist = euclidean_distance(test_point, training_Data[i])
        ds.append((dist, training_labels[i]))
    ds.sort(key=lambda x: x[0])
    kN_labels = [label for _,label in ds[:k]]
    return Counter(kN_labels).most_common(1)[0][0]
    
```

```python
knn_predict(training_data, training_labels, test_point, k)
```
